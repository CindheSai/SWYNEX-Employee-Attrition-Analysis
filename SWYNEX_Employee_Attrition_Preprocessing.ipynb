```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

uploaded = files.upload()
```

```python

import io

csv_files = [
    name for name in uploaded
    if name.lower().endswith(".csv")
]

if not csv_files:
    raise ValueError("Please upload a CSV file.")

df = pd.read_csv(io.BytesIO(uploaded[csv_files[0]]))

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
## Cell 1 — Inspect the dataset structure

# STEP 4.1 — Dataset Structure Inspection

print("=" * 60)
print("EMPLOYEE ATTRITION — DATASET OVERVIEW")
print("=" * 60)

print(f"Number of records : {df.shape[0]:,}")
print(f"Number of columns : {df.shape[1]:,}")
print(f"Memory usage      : {df.memory_usage(deep=True).sum() / 1024:.2f} KB")

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 records:")
display(df.head())

print("\nLast 5 records:")
display(df.tail())

print("\nRandom sample of 5 records:")
display(df.sample(n=min(5, len(df)), random_state=42))
```

```python
## Cell 2 — Inspect column data types and non-null counts


# STEP 4.2 — Schema and Data Types

print("DATASET SCHEMA")
print("-" * 60)

df.info()

schema = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Non_Null_Count": df.notna().sum().values,
    "Missing_Count": df.isna().sum().values,
    "Unique_Values": df.nunique(dropna=True).values
})

display(schema)
```

```python
## Cell 3 — Generate descriptive statistics


# STEP 4.3 — Statistical Summary

print("NUMERICAL FEATURES")
display(df.describe().T.round(2))

print("\nALL FEATURES — INCLUDING CATEGORICAL COLUMNS")
display(df.describe(include="all").T)
```

```python
## Cell 4 — Detect missing values


# STEP 4.4 — Missing-Value Audit

missing_audit = pd.DataFrame({
    "Missing_Count": df.isna().sum(),
    "Missing_Percentage": (df.isna().mean() * 100).round(2),
    "Data_Type": df.dtypes.astype(str)
})

missing_audit = missing_audit.sort_values(
    by="Missing_Count",
    ascending=False
)

display(missing_audit)

total_missing = int(df.isna().sum().sum())
total_cells = int(df.shape[0] * df.shape[1])

print(f"Total missing cells: {total_missing:,}")
print(f"Total dataset cells: {total_cells:,}")
print(f"Overall missing percentage: {total_missing / total_cells * 100:.2f}%")
```

```python
## Cell 5 — Detect duplicate records


# STEP 4.5 — Duplicate Record Audit

duplicate_count = int(df.duplicated().sum())
duplicate_percentage = (
    duplicate_count / len(df) * 100 if len(df) else 0
)

print(f"Exact duplicate rows: {duplicate_count}")
print(f"Duplicate percentage: {duplicate_percentage:.2f}%")

if duplicate_count > 0:
    print("\nPreview of duplicate rows:")
    display(df.loc[df.duplicated(keep=False)].sort_values(
        by=df.columns.tolist()
    ).head(10))
else:
    print("No exact duplicate rows detected.")
```

```python
## Cell 6 — Analyze the attrition target


# STEP 4.6 — Target Variable Audit

if "Attrition" not in df.columns:
    raise ValueError(
        "The required 'Attrition' column was not found."
    )

print("Target value counts:")
display(df["Attrition"].value_counts(dropna=False).to_frame("Count"))

print("\nTarget distribution (%):")
display(
    (
        df["Attrition"].value_counts(normalize=True, dropna=False)
        * 100
    ).round(2).to_frame("Percentage")
)

print("\nMissing target values:", int(df["Attrition"].isna().sum()))
print("Unique target labels:", df["Attrition"].dropna().unique().tolist())
```

```python
## Cell 7 — Visualize target distribution


# STEP 4.7 — Attrition Distribution Visualization

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))

ax = sns.countplot(
    data=df,
    x="Attrition",
    order=df["Attrition"].value_counts().index
)

ax.set_title("Employee Attrition Distribution", fontsize=15)
ax.set_xlabel("Attrition Status")
ax.set_ylabel("Number of Employees")

for container in ax.containers:
    ax.bar_label(container, fmt="%.0f")

plt.tight_layout()
plt.show()
```

```python

# SWYNEX Task 1 — Consolidated Data Quality Report

print("=" * 60)
print("EMPLOYEE ATTRITION: DATA QUALITY REPORT")
print("=" * 60)

print(f"Dataset shape: {df.shape}")
print(f"Total missing cells: {df.isna().sum().sum()}")
print(f"Exact duplicate rows: {df.duplicated().sum()}")
print(f"Duplicate column names: {df.columns.duplicated().sum()}")

print("\nColumn data types:")
display(df.dtypes.astype(str).value_counts().to_frame("Column Count"))

print("\nMissing values by column:")
missing = df.isna().sum()
display(missing[missing > 0].to_frame("Missing Count"))

print("\nAttrition distribution:")
display(
    df["Attrition"]
    .value_counts(dropna=False)
    .to_frame("Employee Count")
)

print("\nAttrition percentage:")
display(
    (df["Attrition"].value_counts(normalize=True, dropna=False) * 100)
    .round(2)
    .to_frame("Percentage")
)
```

```python

# SWYNEX Task 1 — Attrition Rate Analysis

required_columns = ["Attrition", "OverTime", "Department", "JobRole"]
missing_columns = [c for c in required_columns if c not in df.columns]

if missing_columns:
    raise ValueError(f"Missing expected columns: {missing_columns}")

eda_df = df.copy()
eda_df["Attrition_Flag"] = (
    eda_df["Attrition"].astype(str).str.strip().eq("Yes").astype(int)
)

for column in ["OverTime", "Department", "JobRole"]:
    summary = (
        eda_df.groupby(column, dropna=False)
        .agg(
            Employees=("Attrition_Flag", "size"),
            Attrition_Count=("Attrition_Flag", "sum"),
            Attrition_Rate=("Attrition_Flag", "mean")
        )
        .sort_values("Attrition_Rate", ascending=False)
    )

    summary["Attrition_Rate"] = (
        summary["Attrition_Rate"] * 100
    ).round(2)

    print(f"\nAttrition analysis by {column}")
    display(summary)
```

```python

# Attrition Rate by Overtime Status

overtime_summary = (
    eda_df.groupby("OverTime", dropna=False)["Attrition_Flag"]
    .mean()
    .mul(100)
    .reset_index(name="Attrition_Rate")
)

plt.figure(figsize=(8, 5))

ax = sns.barplot(
    data=overtime_summary,
    x="OverTime",
    y="Attrition_Rate"
)

ax.set_title("Employee Attrition Rate by Overtime Status")
ax.set_xlabel("Overtime")
ax.set_ylabel("Employees with Attrition = Yes (%)")

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%")

plt.tight_layout()
plt.show()
```

```python
# STEP 6: Analyze employee attrition by key factors

required_columns = ["Attrition", "OverTime", "Department", "JobRole"]
missing_columns = [col for col in required_columns if col not in df.columns]

if missing_columns:
    raise ValueError(f"Missing expected columns: {missing_columns}")

eda_df = df.copy()

# Convert the target into a binary indicator: Yes = 1, No = 0
eda_df["Attrition_Flag"] = (
    eda_df["Attrition"].astype(str).str.strip().eq("Yes").astype(int)
)

# Analyze attrition across important employee categories
for column in ["OverTime", "Department", "JobRole"]:
    summary = (
        eda_df.groupby(column, dropna=False)
        .agg(
            Employees=("Attrition_Flag", "size"),
            Attrition_Count=("Attrition_Flag", "sum"),
            Attrition_Rate=("Attrition_Flag", "mean")
        )
        .sort_values("Attrition_Rate", ascending=False)
    )

    summary["Attrition_Rate"] = (
        summary["Attrition_Rate"] * 100
    ).round(2)

    print(f"\nAttrition Analysis by {column}")
    display(summary)
```

```python

# STEP 7: Attrition visualization dashboard

import matplotlib.pyplot as plt
import seaborn as sns

# Confirm the required data is available
if "Attrition_Flag" not in eda_df.columns:
    eda_df["Attrition_Flag"] = (
        eda_df["Attrition"].astype(str).str.strip().eq("Yes").astype(int)
    )

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Chart 1: Attrition class distribution
sns.countplot(
    data=eda_df,
    x="Attrition",
    order=["No", "Yes"],
    hue="Attrition",
    legend=False,
    ax=axes[0]
)
axes[0].set_title("Employee Attrition Distribution")
axes[0].set_xlabel("Attrition")
axes[0].set_ylabel("Number of Employees")

# Chart 2: Attrition rate by overtime
overtime_summary = (
    eda_df.groupby("OverTime", dropna=False)["Attrition_Flag"]
    .mean()
    .mul(100)
    .reset_index(name="Attrition_Rate")
)

sns.barplot(
    data=overtime_summary,
    x="OverTime",
    y="Attrition_Rate",
    hue="OverTime",
    legend=False,
    ax=axes[1]
)
axes[1].set_title("Attrition Rate by Overtime")
axes[1].set_xlabel("Overtime Status")
axes[1].set_ylabel("Attrition Rate (%)")

for container in axes[1].containers:
    axes[1].bar_label(container, fmt="%.1f%%")

plt.tight_layout()
plt.show()
```
